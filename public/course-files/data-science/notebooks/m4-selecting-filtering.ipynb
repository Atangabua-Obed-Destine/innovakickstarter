## Part 2: The toolkit (run and read each example)
- One column: `df["Branch"]` (a Series). Several columns: `df[["Branch", "TotalSale_FCFA"]]` (a DataFrame; note the double brackets).
- A **mask** is a True/False Series: `df["Branch"] == "Nkwen"`. Filter with `df[mask]`.
- Combine masks with `&` (and), `|` (or), `~` (not). **Put each condition in parentheses.** Do not use the words `and`/`or`.
- `.isin([...])` for "one of these values"; `.between(a, b)` for a range (both ends included).
- `.loc[rows, columns]` selects **by label or mask**: `df.loc[mask, ["Date", "Product"]]`.
- `.iloc[rows, columns]` selects **by position**: `df.iloc[0:5, 0:3]` is the first 5 rows and first 3 columns.
- `.sort_values("col", ascending=False)` sorts; `len(result)` or `result.shape[0]` counts rows.
- If you plan to change a filtered table later, add `.copy()` so you work on your own copy.
