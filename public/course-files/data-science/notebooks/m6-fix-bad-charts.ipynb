A junior colleague made four charts of Nkwen Traders' January-June 2026 sales for a management meeting. Each one is wrong or misleading in at least one way.

```python
# Setup: run this cell first
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

sales = pd.read_csv("https://innovakickstarter.com/course-files/data-science/data/nkwen_traders_ch2_sales.csv")
sales["Date"] = pd.to_datetime(sales["Date"], format="%Y-%m-%d")
sales["Month"] = sales["Date"].dt.to_period("M")

# One colour per branch, used in every chart so readers learn it once
BRANCH_COLORS = {"Nkwen": "#1b6ca8", "Mile 4": "#e07b00", "Commercial Avenue": "#2a9d55", "Bafoussam Centre": "#b03a48"}
DAY_ORDER = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
fcfa = mticker.StrMethodFormatter("{x:,.0f}")   # 1500000 -> 1,500,000

print(sales.shape)
sales.head()
```

## The four bad charts
Run each cell and look closely. Do not change these cells: they are the evidence.

```python
# BAD CHART 1
bamenda = sales[sales["Branch"] != "Bafoussam Centre"].groupby("Branch")["TotalSale_FCFA"].sum().sort_values()
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(bamenda.index, bamenda.values, color=["grey", "grey", "red"])
ax.set_ylim(14_000_000, 15_600_000)
ax.set_title("Commercial Avenue CRUSHES Nkwen!")
plt.show()
```

```python
# BAD CHART 2
by_product = sales.groupby("Product")["TotalSale_FCFA"].sum()
fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(by_product.values, labels=by_product.index)
ax.set_title("Revenue")
plt.show()
```

```python
# BAD CHART 3
avg_by_category = sales.groupby("Category")["TotalSale_FCFA"].mean()
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(avg_by_category.index, avg_by_category.values)
plt.show()
```

```python
# BAD CHART 4
mile4 = sales[sales["Branch"] == "Mile 4"].groupby("Month")["TotalSale_FCFA"].sum()
feb_mar = mile4.loc[pd.Period("2026-02", freq="M"):pd.Period("2026-03", freq="M")]
fig, ax = plt.subplots(figsize=(5, 4))
ax.bar(feb_mar.index.astype(str), feb_mar.values, color=["grey", "green"])
growth = (feb_mar.iloc[1] / feb_mar.iloc[0] - 1) * 100
ax.set_title(f"Mile 4 is booming: +{growth:.0f}%!")
ax.yaxis.set_major_formatter(fcfa)
plt.show()
```
