### Step 5: A note on pandas (read, then answer)
You may read that `map()` is "what pandas does under the hood". That is **not accurate**. Most pandas operations (`df["price"] * 1.05`, `df[df["quantity"] < 10]`) are **vectorised**: they work on a whole column at once in fast compiled code, without a Python loop. `.apply()` with a lambda does call your Python function once per row, which is much slower on big data. In Milestone 4 you will prefer vectorised code and keep `lambda` for small jobs such as `sorted(..., key=lambda pair: pair[1])`.
