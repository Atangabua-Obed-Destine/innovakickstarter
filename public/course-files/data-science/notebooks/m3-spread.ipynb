### Note: the same idea, different defaults in different tools
| Tool | Default | Version |
|---|---|---|
| pandas `df["col"].std()` | `ddof=1` | sample (divides by n - 1) |
| NumPy `np.std(arr)` | `ddof=0` | population (divides by n) |
| Python `statistics.stdev` / `statistics.pstdev` | | sample / population |

`ddof` means "delta degrees of freedom": the divisor is `n - ddof`. In this track we use the **sample** version by default, so your numbers match pandas later. If your answer differs a little from a classmate's, check `ddof` first.
