````python
# Setup: run this cell first
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

sales = pd.read_csv("https://innovakickstarter.com/course-files/data-science/data/nkwen_traders_ch2_sales.csv")
sales["Date"] = pd.to_datetime(sales["Date"], format="%Y-%m-%d")
sales["Month"] = sales["Date"].dt.to_period("M")

# One colour per branch, used in every chart so readers learn it once
BRANCH_COLORS = {"Nkwen": "#1b6ca8", "Mile 4": "#e07b00", "Commercial Avenue": "#2a9d55", "Bafoussam Centre": "#b03a48"}
DAY_ORDER = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
fcfa = mticker.StrMethodFormatter("{x:,.0f}")   # 1500000 -> 1,500,000

print(sales.shape)
sales.head()
````

## Step 1. The pattern you will use for every chart
Matplotlib has two styles. The **pyplot style** (`plt.plot(...)`, `plt.title(...)`) is fine for quick checks, but it hides which chart you are changing. The **object-oriented style** names the figure and the axes, so you always know what you change. Use it for all your work:

```python
fig, ax = plt.subplots(figsize=(8, 4))   # fig = the whole image, ax = one chart area
ax.plot(x, y)                            # draw on THIS chart
ax.set_title("...")                      # label THIS chart
ax.set_xlabel("..."); ax.set_ylabel("...")
fig.savefig("name.png", dpi=150, bbox_inches="tight")   # save the whole image
```
Run the worked example below. It is complete: read every line.

````python
# Worked example: number of sales per month (all branches)
count_by_month = sales.groupby("Month").size()

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(count_by_month.index.astype(str), count_by_month.values, marker="o", color="#444444")
ax.set_title("Number of sales per month, all branches, Jan-Jun 2026")
ax.set_xlabel("Month")
ax.set_ylabel("Number of sales")
ax.set_ylim(bottom=0)        # start at zero so the size of changes is honest
ax.grid(axis="y", alpha=0.3)
plt.show()
````
