## Key ideas
- **Probability from data** = (number of rows where the event happens) / (number of rows). Always between 0 and 1.
- **Complement:** P(not A) = 1 - P(A).
- **And:** P(A and B) = rows where both happen / all rows.
- **Conditional:** P(A | B), "A given B": first narrow the rows to those where B happened, then ask what share of *those* rows also have A. The denominator is the narrowed group.
- **Independence:** A and B are independent if knowing B does not change the chance of A, i.e. P(A | B) is about the same as P(A).
- **Multiplication rule (independent events only):** P(A and B) = P(A) x P(B).
