## Step 5: Apply the vocabulary, and check with mean vs median
- **Roughly symmetric (bell-shaped, like a normal distribution):** one peak in the middle, tails about equal on both sides. Mean and median are close.
- **Right-skewed:** one peak on the left, a long tail to the right. The mean is pulled above the median.
- **Left-skewed:** the mirror image. The mean is below the median.
- **Bimodal:** two separate peaks. Often two different groups mixed together.
