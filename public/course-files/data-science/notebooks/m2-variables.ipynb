### Step 2: Why FCFA prices are `int`, not `float`
FCFA amounts are whole numbers (there is no "half franc" on a receipt). Computers store floats in binary, so some decimals cannot be stored exactly. Run the cell below and read the result.

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

Tiny errors like this can add up across thousands of sales and make totals that do not match. **Rule for this track: store FCFA amounts as `int`.**

Arithmetic operators: `+` `-` `*` `/` `//` `%` `**`. Note that `/` always gives a float (`7000 / 2` is `3500.0`), while `//` divides and keeps a whole number (`7000 // 2` is `3500`). `%` gives the remainder.
