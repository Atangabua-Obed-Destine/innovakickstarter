## How groupby works: split, apply, combine
`df.groupby("Branch")["TotalSale_FCFA"].sum()`
1. **Split** the rows into one group per branch.
2. **Apply** `sum()` to the `TotalSale_FCFA` column of each group.
3. **Combine** the results into one Series indexed by branch.

Several statistics at once: `.agg(["sum", "count", "mean"])`. Named results:
```python
df.groupby("Branch").agg(
    revenue=("TotalSale_FCFA", "sum"),
    transactions=("TransactionID", "count"),
)
```
